```python
# 📦 Cell 1: 기본 라이브러리 및 환경 설정
# ========================================
import sys
import os
from pathlib import Path

# 현재 디렉토리를 Python 경로에 추가
current_dir = Path.cwd()
if str(current_dir) not in sys.path:
    sys.path.insert(0, str(current_dir))

# 상위 디렉토리(module 폴더가 있는 위치)를 Python 경로에 추가
parent_dir = current_dir.parent  # calendar format re copy
if str(parent_dir) not in sys.path:
    sys.path.insert(0, str(parent_dir))

# module 폴더 찾기 (현재 디렉토리 또는 상위 디렉토리)
module_dir = None

# 1. 현재 디렉토리에서 module 폴더 확인
current_module_dir = current_dir / "module"
if current_module_dir.exists():
    module_dir = current_module_dir
    print(f"✅ module 폴더 발견 (현재 디렉토리): {module_dir}")
elif (parent_dir / "module").exists():
    # 2. 상위 디렉토리에서 module 폴더 확인
    module_dir = parent_dir / "module"
    print(f"✅ module 폴더 발견 (상위 디렉토리): {module_dir}")
    # module 폴더의 부모 디렉토리가 sys.path에 있는지 확인
    if str(module_dir.parent) not in sys.path:
        sys.path.insert(0, str(module_dir.parent))
else:
    print(f"⚠️ module 폴더를 찾을 수 없습니다.")
    print(f"   현재 디렉토리: {current_module_dir}")
    print(f"   상위 디렉토리: {parent_dir / 'module'}")

# 환경 변수 로드
from dotenv import load_dotenv
load_dotenv()

# LangChain 및 AI 모델 관련 라이브러리
from langchain_core.runnables import RunnableConfig
from langchain_teddynote.messages import random_uuid

print("✅ 기본 라이브러리 로드 완료")
print(f"📁 현재 작업 디렉토리: {current_dir}")
print(f"📁 Python 경로에 추가된 디렉토리: {parent_dir}")
if module_dir:
    print(f"📁 module 폴더 위치: {module_dir}")
print(f"📁 sys.path 확인 (상위 5개):")
for i, path in enumerate(sys.path[:5], 1):
    print(f"   {i}. {path}")
```

```python
# 🎯 Cell 6: 질문 실행 및 결과 확인
# ========================================
print("=" * 80)
print("🎯 RAG 시스템 실행")
print("=" * 80)

# 테스트 질문
test_questions = [
    {
        "question": "작물에 병이 생겼는데 처방법을 알려주세요",
        "image": "angular_leafspot351.jpg"
    }
]

# 또는 사용자 입력 받기
# question = input("\n질문을 입력하세요: ").strip()

for i, test_case in enumerate(test_questions, 1):
    question = test_case["question"]
    image_path = test_case.get("image")

    print(f"\n📝 테스트 {i}/{len(test_questions)}")
    print(f"질문: {question}")
    if image_path:
        print(f"🖼️ 이미지: {image_path}")
    else:
        print("🖼️ 이미지: 없음")
    print("-" * 80)

    if image_path:
        if os.path.exists(image_path):
            print(f"✅ 이미지 파일 확인됨: {image_path}")
        else:
            print(f"❌ 이미지 파일 확인 실패: {image_path}")
            print(f"이미지 없이 진행합니다.")
            image_path = None

    # RAG 시스템 실행
    result = run_rag_system(
        question=question,
        image_path=image_path,
        config={"app": rag_app}  # 초기화된 앱 사용
    )
    
    # 결과 상세 확인
    print("\n" + "=" * 80)
    print("📊 상세 결과:")
    print("=" * 80)
    
    # 답변
    answer = result.get("answer", result.get("generation", ""))
    print(f"\n💬 답변:")
    print(answer)

    # 질문 유효성
    question_valid = result.get("question_valid", True)
    print(f"\n✅ 질문 유효성: {'유효' if question_valid else '무효'}")
    
    # 복잡도 수준
    complexity_level = result.get("complexity_level", "unknown")
    print(f"📊 복잡도 수준: {complexity_level}")
    
    # 품질 점수
    # 품질 점수
    quality_scores = result.get("quality_scores", {})
    if quality_scores:
        print(f"\n📊 RAG 품질 점수:")
        # 평가 방식 정보 추가
        evaluation_method = quality_scores.get('evaluation_method', 'Unknown')
        print(f"   - 평가 방식: {evaluation_method}")  # 추가됨
        print(f"   - 검색 정확도: {quality_scores.get('retrieval_accuracy', 0):.3f}")
        print(f"   - 답변 관련성: {quality_scores.get('answer_relevance', 0):.3f}")
        print(f"   - 답변 정확도: {quality_scores.get('answer_correctness', 0):.3f}")
        print(f"   - 할루시네이션 점수: {quality_scores.get('hallucination_score', 0):.3f}")
        print(f"   - 전체 점수: {quality_scores.get('overall_score', 0):.3f}")
    
    # LLM Judge 점수
    llm_judge_scores = result.get("llm_judge_scores", {})
    if llm_judge_scores:
        print(f"\n📊 LLM Judge 점수:")
        print(f"   - 정확성: {llm_judge_scores.get('accuracy', 0)}/100")
        print(f"   - 완성도: {llm_judge_scores.get('completeness', 0)}/100")
        print(f"   - 논리성: {llm_judge_scores.get('logical_consistency', 0)}/100")
        print(f"   - 유용성: {llm_judge_scores.get('usefulness', 0)}/100")
        print(f"   - 종합 점수: {llm_judge_scores.get('overall_score', 0)}/100")
        print(f"   - 유효성: {'유효' if llm_judge_scores.get('is_valid', False) else '무효'}")
    
    # 검색된 문서 정보
    documents = result.get("retrieved_docs", result.get("documents", []))
    if documents:
        print(f"\n📚 검색된 문서 ({len(documents)}개):")
        for j, doc in enumerate(documents[:5], 1):  # 상위 5개만 출력
            source = doc.metadata.get('source', 'Unknown')
            crop = doc.metadata.get('crop', 'Unknown')
            page = doc.metadata.get('page', 'Unknown')
            print(f"   {j}. {source}")
            print(f"      작물: {crop}, 페이지: {page}")
            print(f"      내용 미리보기: {doc.page_content[:100]}...")
            print()
    
    # 상태 정보
    status = result.get("status", "unknown")
    print(f"📊 처리 상태: {status}")
    
    if i < len(test_questions):
        print("\n" + "=" * 80)

print("\n🎉 모든 테스트 완료!")
print("=" * 80)
```
